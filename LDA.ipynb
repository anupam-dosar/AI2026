```python
import pandas as pd
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
import gensim
import gensim.corpora as corpora
```

```python
data = pd.read_csv('news_articles.csv')
```

```python
data.head()
```

```python
data.info()
```

```python
articles = data['content']
```

```python
articles = articles.str.lower().apply(lambda x: re.sub(r"[^\w\s]", "", x))
```

```python
en_stopwords = stopwords.words('english')
articles = articles.apply(lambda row: ' '.join(w for w in row.split() if w not in en_stopwords))
```

```python
articles.head()
```

```python
articles = articles.apply(lambda x: word_tokenize(x))
```

```python
articles.head()
```

```python
ps = PorterStemmer()
articles = articles.apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
articles
```

```python
dictionary= corpora.Dictionary(articles)
print(dictionary)
```

```python
doc_term = [dictionary.doc2bow(text) for text in articles]
```

```python
print(doc_term)
```

```python
num_topics =10
```

```python
lda_model = gensim.models.LdaModel(corpus=doc_term, id2word=dictionary, num_topics=num_topics)
```

```python
lda_model.print_topics(num_topics=num_topics, num_words=10)
```
