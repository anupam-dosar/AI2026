```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import re
import pandas as pd

data = pd.read_csv('tripadvisor_hotel_reviews.csv')

data.info()
```

```python
data.head()
```

```python
data['Review'][0]
```

```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
data.head()
```

```python
en_stopwords = stopwords.words('english')
en_stopwords.remove('not')
```

```python
data['review_no_stopw'] = data['review_lowercase'].apply(lambda x: ' '.join(word for word in x.split() if word not in (en_stopwords)))
```

```python
data.head()
```

```python
data['review_no_stopw'][0]
```

```python
data['review_no_stopw_no_punct'] = data.apply(lambda x: re.sub(r'[*]', 'star', x['review_no_stopw']), axis=1)
```

```python
data.head()
```

```python
data['review_no_stopw_no_punct'] = data.apply(lambda x: re.sub(r'[^\w\s]', '', x['review_no_stopw_no_punct']), axis=1)
```

```python
data.head()
```

```python
data['review_no_stopw_no_punct'][0]
```

```python
data['tokenized'] = data.apply(lambda x: word_tokenize(x['review_no_stopw_no_punct']), axis=1)
```

```python
data.head()
```

```python
data['tokenized'][0]
```

```python
ps = PorterStemmer()
```

```python
data['stemmed'] = data['tokenized'].apply(lambda tokens: ps.stem(token) for token in tokens)
```
